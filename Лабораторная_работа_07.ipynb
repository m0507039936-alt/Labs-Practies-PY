# Мое решение

````python
F = open("p7_data_05.csv")
#В VS Code все работает. В гугл колабе с файлом какая-то проблема. Ошибку не выдает, но файл не считывает. Надеюсь у вас все открылось


header = F.readline().strip().split(',')
print(header)

data_tickets = []
for line in F:
  line = line.strip().split(',')
  person = {}
  for i, key in enumerate(header):
    person[key] = line[i]
  data_tickets.append(person)

F.close()

st_tick = []
rs_tick = []
vr_tick = []
ms_tick = []

for ticket in data_tickets:
  if ticket["City_From"] == "Stavropol":
    st_tick.append(float(ticket["Price"][1:]))
  elif ticket["City_From"] == "Moscow":
    ms_tick.append(float(ticket["Price"][1:]))
  elif ticket["City_From"] == "Rostov-on-Don":
    rs_tick.append(float(ticket["Price"][1:]))
  elif ticket["City_From"] == "Voronezh":
    vr_tick.append(float(ticket["Price"][1:]))


print(f"Количество билетов с пунктом отправления из Ставрополя: {len(st_tick)}")
print(f"Средняя цена билета для пункта отправления Ставрополь: {sum(st_tick)/len(st_tick)}")
print(f"Средняя цена билета для пункта отправления Воронеж: {sum(vr_tick)/len(vr_tick)}")
print(f"Средняя цена билета для пункта отправления Ростов: {sum(rs_tick)/len(rs_tick)}")
print(f"Средняя цена билета для пункта отправления Москва: {sum(ms_tick)/len(ms_tick)}")

men_cnt = 0
women_cnt = 0
for ticket in data_tickets:
  if ticket["Gender"] == "M":
    men_cnt += 1
  else:
    women_cnt += 1
print(f"Билеты купленные мужчинами: {men_cnt}")
print(f"Билеты купленные женщинами: {women_cnt}")

thesame_cnt = 0
for ticket in data_tickets:
  if ticket["City_From"] == ticket["City_To"]:
    thesame_cnt += 1
print(f"Билеты с один. пунтками отправления и прибытия: {thesame_cnt}")
````
