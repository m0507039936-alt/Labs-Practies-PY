Файл pr6.txt расположен репозитории. Выполним копирование файла в локальное хранилище Google Colab, используя следующую команду:

````python
!wget https://raw.githubusercontent.com/enikolaev/IT-and-Programming/refs/heads/main/data/pract6/pr6_1.txt
````

````python
#Эт мое решение :)
f = open("pr6_1.txt").read().strip()

import string as st
symbols = st.ascii_uppercase + st.digits
print(symbols)

max_text = ""
r = ""
for i in symbols:
    text = f
    text = text.replace(f"{i}", "*")
    text = [el for el in text.split("*")]
    for element in text:
      if len(element) > len(max_text):
        max_text = element
        t = i
print(max_text, len(max_text), t)
````
