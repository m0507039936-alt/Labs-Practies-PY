## Решение задачи


````python
f = open("data_v1.txt")
d = f.read().strip()

max_posl = ""
posl = ""

for el in d:
    if not(el.isupper()) and el != "3" and el != "6" and el !="9":
        posl += el
    else:
        posl = ""
    if len(posl) > len(max_posl):
                max_posl = posl
print(max_posl, len(max_posl))
````
